## What does sound speed depend on?

- Temperature, salinity and pressure -- and their very different weights.
- What the two correlations below show, and why the temperature one is so close to 1 in this record specifically.
- A caution: a high correlation here is partly a property of this narrow temperature range, not a universal law.

```python
corr_ss_PSAL = xr.corr(ds.PSAL, ds.SVEL)
corr_ss_TEMP = xr.corr(ds.TEMP, ds.SVEL)
```

```python
print(f'Correlation, sound speed and PSAL: {corr_ss_PSAL:.4f}')
print(f'Correlation, sound speed and TEMP: {corr_ss_TEMP:.4f}')
```
