```python
from kval.data import moored
import xarray as xr
%matplotlib widget
```

```python
ds = moored.load_moored('../../data/moored_CTD_test_data/intermediate_data/AT200_21_22_SBE37_15252_113m_from_raw.nc')
```

```python
ds = moored.calculate_PSAL(ds)
```

```python
ds = moored.calculate_ss(ds)
```

```python
ds
```

```python
moored.plot(ds)
```

### What does sound speed depend on?

- 

```python
corr_ss_PSAL = xr.corr(ds.PSAL, ds.SVEL)
corr_ss_TEMP = xr.corr(ds.TEMP, ds.SVEL)
```

```python
print(f'Correlation, sound speed and PSAL: {corr_ss_PSAL:.3f}')
print(f'Correlation, sound speed and TEMP: {corr_ss_TEMP:.3f}')
```
